```python
# Listen to 8 random segments with their text: do audio and text agree?
import pandas as pd, IPython.display as ipd
m = pd.read_csv(f"{OUT}/metadata.tsv", sep="\t", quoting=3)
print(m.groupby("split").seconds.agg(["count", "sum"]).assign(hours=lambda d: d["sum"] / 3600))
for _, r in m.sample(8, random_state=0).iterrows():
    print(f"[{r.split}] {r.source_id}: {r.text}")
    display(ipd.Audio(f"{OUT}/{r.file}"))
```
